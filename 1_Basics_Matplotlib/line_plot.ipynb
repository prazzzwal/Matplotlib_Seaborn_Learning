# Line Plot
- used mainly for Bivariate Analysis
- syntax: <br>
  plt.plot(categorical,numerial)

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
price = [2510,2561,6657,3126,9138,1345,4513]
year = [2010,2011,2012,2013,2014,2015,2016]

plt.plot(year,price) #plt.plot(x-axis,y-axis)
```

```python
data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
              "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"],
    "Marketing_Spend": [20, 25, 30, 22, 35, 40, 38, 28, 26, 32, 45, 50],
    "Website_Visitors": [2000, 2500, 3200, 2700, 4000, 4500, 4200, 3100, 2900, 3600, 5000, 5500],
    "Units_Sold": [150, 180, 210, 170, 260, 300, 280, 200, 190, 230, 340, 380],
    "Revenue": [30, 36, 42, 34, 52, 60, 56, 40, 38, 46, 68, 76]
}

df = pd.DataFrame(data)
df
```

### Plotting a line graph

```python
plt.plot(df['Month'],df['Marketing_Spend'],color='red')
plt.plot(df['Month'],df['Revenue'])

# title for graph

plt.title("Graph of Marketing Spend and Revenue Generated")
plt.xlabel('Month')
plt.ylabel('Revenue in thousands')
```

### Setting linestyle
- dashed
- dotted
- dashdot (dashed followed by dotted)

```python
plt.plot(df['Month'],df['Units_Sold'],color='red',linestyle='dashed')
plt.plot(df['Month'],df['Revenue'],linestyle='dotted')
plt.plot(df['Month'],df['Marketing_Spend'],color='green',linestyle='dashdot')
```

### Setting linewidth

```python
plt.plot(df['Month'],df['Units_Sold'],linestyle='dashed',linewidth=3)
plt.plot(df['Month'],df['Revenue'],linestyle='dotted',linewidth=4)
plt.plot(df['Month'],df['Marketing_Spend'],linestyle='dashdot',linewidth=7)
```

### Manipulating markers 

```python
plt.plot(df['Month'],df['Units_Sold'],linewidth=2,marker='+',markersize=10)
plt.plot(df['Month'],df['Revenue'],linewidth=2,marker='o')
plt.plot(df['Month'],df['Marketing_Spend'],linewidth=2,marker='D',markersize=8)
```

### Setting Legend
- first set label then use the syntax below
- plt.legend()

```python
plt.plot(df['Month'],df['Units_Sold'],linewidth=2,marker='+',markersize=10,label='Units Sold')
plt.plot(df['Month'],df['Revenue'],linewidth=2,marker='o',label='Revenue')
plt.plot(df['Month'],df['Marketing_Spend'],linewidth=2,marker='D',markersize=8,label='Marketing Spend')

plt.legend() #default location is best
```

### Limiting axes
- used whenever the curve starts looking flat due to the outlier 

```python
price = [2510,2561,6657,3126,9138,1345,4513,5000000]
year = [2010,2011,2012,2013,2014,2015,2016,2017]

plt.plot(year,price)

#setting limit for y axis
plt.ylim(0,50000)

#setting limit for x axis
plt.xlim(2012,2015)
```

### Setting Grid

```python
plt.plot(df['Month'],df['Units_Sold'],linewidth=2,marker='o',markersize=6,label='Units Sold')

plt.grid()

plt.show()
```
