# Bar Plot



```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
np.random.seed(10)

teams = ["CSK", "MI", "RCB", "KKR", "RR", "SRH", "DC", "PBKS", "GT", "LSG"]

df = pd.DataFrame({
    "team": teams,
    "matches_played": np.random.randint(12, 18, len(teams)),
    "wins": np.random.randint(6, 14, len(teams)),
    "losses": np.random.randint(4, 10, len(teams)),
    "runs_scored": np.random.randint(1600, 2500, len(teams)),
    "wickets_taken": np.random.randint(50, 120, len(teams))
})

df
```

### Vertical Barplot

```python
plt.bar(df['team'],df['wins'])
```

### Horizontal Barplot

```python
plt.barh(df['team'],df['runs_scored'])
```

### Creating multiple bar plots

#### Creating Manually

```python
space = np.arange(df.shape[0])

plt.bar(space - 0.2,df['matches_played'],width=0.2)
plt.bar(space,df['wins'],width=0.2)
plt.bar(space + 0.2,df['losses'],width=0.2)

plt.xticks(space,df['team']) #providing labels to the x-axis 
plt.show()

```

#### Using plt.bar()

```python
df.set_index("team")[["wins", "losses"]].plot(kind="bar", figsize=(10,6))
plt.ylabel("matches")
plt.title("Wins vs Losses per Team")
plt.show()
```

### Stacked Bar Plot

```python

data = pd.DataFrame({
    "Quarter": ["Q1", "Q2", "Q3", "Q4"],
    "Electronics": [120, 150, 180, 160],
    "Clothing": [80, 90, 100, 110],
    "Groceries": [60, 70, 75, 85]
})

data
```

```python
plt.bar(data['Quarter'],data['Electronics'],label='Electronics')
plt.bar(data['Quarter'],data['Clothing'],bottom=data['Electronics'],label='Clothing')
plt.bar(data['Quarter'],data['Groceries'],bottom=(data['Electronics'] + data['Clothing']),label='Groceries')

plt.legend()
plt.show()
```
